# RQ2 — Which algorithm is most reliable in terms of low overfitting?
**Metric**: `overfitting = change_perc_validate − change_perc_train` (more positive = more overfitting; negative = generalises better than it trains).  
**Algorithms**: FirstImprovement (FI), GeneticProgrammingUniformConcat (GP), RandomSearch, UMDAAlgorithm (UMDA4GI) — all without LLM.  
**Data**: Perf\<Cycles\>, 5-fold CV, budgets {100, 250, 500, 750, 1000}.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from pathlib import Path

os.makedirs('plots', exist_ok=True)

# READ-ONLY — never write back to results.csv
CSV_PATH = Path('..') / 'results' / 'results.csv'
df_all = pd.read_csv(CSV_PATH)

RQ2_ALGOS = ['FirstImprovement', 'GeneticProgrammingUniformConcat', 'RandomSearch', 'UMDAAlgorithm']

df = df_all[
    (df_all['fitness'] == 'Perf<Cycles>') &
    (df_all['algorithm'].isin(RQ2_ALGOS))
].copy()

# Keep only rows without LLM (pandas reads 'None' from CSV as pd.NA)
df = df[df['llm_model'].isna()]

df['pop_size']             = pd.to_numeric(df['pop_size'],  errors='coerce').fillna(-1).astype(int)
df['max_evals']            = pd.to_numeric(df['max_evals'], errors='coerce').astype(int)
df['change_perc_validate'] = pd.to_numeric(df['change_perc_validate'], errors='coerce')
df['change_perc_train']    = pd.to_numeric(df['change_perc_train'],    errors='coerce')
df['Vt']                   = pd.to_numeric(df['Vt'],                   errors='coerce')
df['Vc']                   = pd.to_numeric(df['Vc'],                   errors='coerce')

print(f'Rows: {len(df)}')
print(f'Algorithms: {sorted(df["algorithm"].unique())}')
print(f'Softwares:  {sorted(df["software"].unique())}')
print(f'max_evals:  {sorted(df["max_evals"].unique())}')

FIXED_COLORS = {
    'FirstImprovement':                cm.tab10(0),
    'GeneticProgrammingUniformConcat': cm.tab10(1),
    'RandomSearch':                    cm.tab10(2),
    'UMDAAlgorithm':                   cm.tab10(3),
    'UMDA4GI':                         cm.tab10(3),
    'UMDA+llama':                      cm.tab10(4),
    'UMDA+qwen':                       cm.tab10(5),
    'LLM-llama':                       cm.tab10(6),
    'LLM-qwen':                        cm.tab10(7),
}
```

```python
def validate_point_or_zero(s: pd.Series) -> float:
    """Mean across 5 folds; returns 0 if any fold is missing, invalid, or found no patch."""
    s = pd.to_numeric(s, errors='coerce')
    if s.isna().any() or len(s) != 5 or (s == 0.0).any():
        return 0.0
    return float(s.mean())

group_cols = ['software', 'algorithm', 'pop_size', 'max_evals', 'fitness']

averaged_results = (
    df.groupby(group_cols, as_index=False)
      .agg(
          change_perc_train   =('change_perc_train',    validate_point_or_zero),
          change_perc_validate=('change_perc_validate', validate_point_or_zero),
          Vt                  =('Vt',                   'mean'),
          Vc                  =('Vc',                   'mean'),
          n_folds             =('fold',                 'count'),
      )
)

# Sanity check
bad = averaged_results[averaged_results['n_folds'] != 5]
if bad.empty:
    print('OK — all groups have exactly 5 folds')
else:
    print(f'WARNING — {len(bad)} groups do not have 5 folds:')
    display(bad)

# Compute overfitting: positive = overfits, negative = generalises better than trains
averaged_results['overfit'] = averaged_results['change_perc_validate'] - averaged_results['change_perc_train']

print(f'Aggregated rows: {len(averaged_results)}')
```

## 1. Overfitting curves — (validate − train) vs max_evals (per software)

```python
legend_name_map = {
    'FirstImprovement':                'FI',
    'GeneticProgrammingUniformConcat': 'GP',
    'RandomSearch':                    'RandomS',
    'UMDAAlgorithm':                   'UMDA4GI',
}
software_display_map = {
    'minisat_hack': 'MiniSAT-hack',
    'sat4j':        'Sat4J',
    'weka':         'Weka',
    'pngOptim':     'OptimPNG',
}
desired_order = ['minisat_hack', 'sat4j', 'weka', 'pngOptim']

line_styles = ['-', '--', ':', '-.']
markers     = ['o', 's', '^', 'D']

algorithms_all = list(averaged_results['algorithm'].unique())
color_map  = {alg: cm.tab10(i % 10)                for i, alg in enumerate(algorithms_all)}
style_map  = {
    alg: {'linestyle': line_styles[i % len(line_styles)],
          'marker':    markers[i % len(markers)],
          'color':     color_map[alg],
          'label':     legend_name_map.get(alg, alg)}
    for i, alg in enumerate(algorithms_all)
}

softwares = [sw for sw in desired_order if sw in averaged_results['software'].values]

fig, axes = plt.subplots(2, 2, figsize=(16, 10), sharex=False, sharey=False)
axes = axes.flatten()

for ax, software in zip(axes, softwares):
    sw_data = averaged_results[averaged_results['software'] == software]
    for alg in algorithms_all:
        alg_data = sw_data[sw_data['algorithm'] == alg].sort_values('max_evals')
        if alg_data.empty:
            continue
        ax.plot(
            alg_data['max_evals'],
            alg_data['overfit'],
            label=style_map[alg]['label'],
            linestyle=style_map[alg]['linestyle'],
            marker=style_map[alg]['marker'],
            color=style_map[alg]['color'],
            linewidth=2, markersize=6,
        )
    ax.set_title(software_display_map.get(software, software), fontsize=15)
    ax.set_xlabel('Max evaluations (stopping criterion)', fontsize=13)
    ax.set_ylabel('Overfitting (validate − train) %', fontsize=13)
    ax.set_xticks(sorted(sw_data['max_evals'].unique()))
    ax.axhline(y=0, color='k', linestyle='-', linewidth=2)
    ax.grid(True)

for i in range(len(softwares), len(axes)):
    axes[i].axis('off')

legend_handles, legend_labels = [], []
for ax in axes[:len(softwares)]:
    h, l = ax.get_legend_handles_labels()
    if h:
        legend_handles, legend_labels = h, l
        break

if legend_handles:
    fig.legend(legend_handles, legend_labels,
               loc='upper center', bbox_to_anchor=(0.5, 1.02),
               ncol=len(legend_labels), fontsize='xx-large', frameon=False)

plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.savefig('plots/RQ2_overfitting_2x2.svg', format='svg', bbox_inches='tight')
plt.show()
```

## 2. Overfitting tables — mean (validate − train) per software × max_evals

```python
grp_ov = averaged_results.groupby(['software', 'algorithm', 'max_evals'], as_index=False)['overfit'].mean()

for alg in sorted(grp_ov['algorithm'].unique()):
    pivot = (
        grp_ov[grp_ov['algorithm'] == alg]
        .pivot(index='max_evals', columns='software', values='overfit')
        .sort_index()
    )
    pivot.columns = [software_display_map.get(c, c) for c in pivot.columns]
    display(f'Overfit (val − train) for {legend_name_map.get(alg, alg)}:')
    display(pivot.round(4))
```

## 3. Ranking — algorithms by mean overfitting per software (1 = least overfitting)

```python
overall_avg_ov = grp_ov.groupby(['software', 'algorithm'], as_index=False)['overfit'].mean()
overall_avg_ov['rank'] = overall_avg_ov.groupby('software')['overfit'].rank(method='min')

rank_alg_ov = overall_avg_ov.pivot(index='algorithm', columns='software', values='rank').sort_index()
rank_alg_ov.index   = [legend_name_map.get(i, i) for i in rank_alg_ov.index]
rank_alg_ov.columns = [software_display_map.get(c, c) for c in rank_alg_ov.columns]

display('Algorithm ranking by mean overfitting per software (1 = least overfitting):')
display(rank_alg_ov)

# Overall rank (sum of ranks across softwares)
rank_alg_ov['Sum'] = rank_alg_ov.sum(axis=1)
rank_alg_ov['Overall'] = rank_alg_ov['Sum'].rank(method='min').astype(int)
display('Overall ranking (lower sum = more consistently low overfitting):')
display(rank_alg_ov.sort_values('Sum'))
```

## 4. Ranking — max_evals by mean overfitting per algorithm (1 = least overfitting)

```python
avg_budget_ov = grp_ov.groupby(['algorithm', 'max_evals'], as_index=False)['overfit'].mean()
avg_budget_ov['rank'] = avg_budget_ov.groupby('algorithm')['overfit'].rank(method='min')

rank_budget_ov = avg_budget_ov.pivot(index='max_evals', columns='algorithm', values='rank').sort_index()
rank_budget_ov.columns = [legend_name_map.get(c, c) for c in rank_budget_ov.columns]

display('max_evals ranking by mean overfitting per algorithm (1 = least overfitting):')
display(rank_budget_ov)
```

## RQ2 — Conclusiones

### Respuesta directa

**UMDA4GI es el algoritmo con menor overfitting global** (Sum de rangos = 7, 1º en MiniSAT-hack, OptimPNG y Weka). En MiniSAT-hack con presupuestos bajos incluso muestra **overfitting negativo** (generaliza mejor de lo que entrena: −10.6% a budget=100, −3.3% a budget=250), lo cual es un resultado notable.

### Ranking global (suma de rangos por software, 1 = menos overfitting)

| Algoritmo | MiniSAT-hack | OptimPNG | Sat4J | Weka | **Sum** |
|-----------|:---:|:---:|:---:|:---:|:---:|
| **UMDA4GI** | **1** | **1** | 4 | **1** | **7** |
| FI | 3 | 2 | 2 | 2 | 9 |
| GP | 4 | 4 | **1** | 3 | 12 |
| RandomS | 2 | 3 | 3 | 4 | 12 |

### Hallazgos por software

**MiniSAT-hack** — único software con overfitting significativo:
- UMDA4GI: **overfitting negativo a presupuestos bajos** (−10.6% a budget=100, −3.3% a budget=250). A presupuestos altos el overfitting crece (24.2% a 500, 31.8% a 1000).
- FI: overfitting alto y variable (41%–105% según budget). Pico en budget=250 (105%).
- GP: **outlier severo a budget=500** (183.9% de overfitting). El resto de budgets oscilan entre 22%–42%. Sugiere un caso de sobreajuste catastrófico puntual.
- RandomSearch: overfitting moderado a presupuestos extremos (100 y 750–1000: ~32%), más bajo en el rango medio (250–500: ~6%).

**Sat4J, Weka, OptimPNG** — overfitting prácticamente nulo:
- Todos los algoritmos muestran overfitting cercano a 0% en estos tres softwares. Las diferencias de ranking son mínimas y sin relevancia práctica.
- UMDA4GI gana en OptimPNG y Weka; GP gana en Sat4J.

### Efecto del presupuesto en el overfitting

No hay un patrón universal. En MiniSAT-hack, UMDA4GI pasa de overfitting negativo (generaliza) a positivo (sobreajusta) al aumentar el presupuesto, lo que sugiere que con pocos evals construye distribuciones que generalizan bien, pero con más evals empieza a explotar características del train. FI y GP muestran alta variabilidad. RandomSearch tiene un patrón en U (menos overfitting en presupuestos medios).

### Implicaciones para el paper

- **UMDA4GI combina mejor mejora (RQ1) con menor overfitting (RQ2)**, lo que lo posiciona como el algoritmo más equilibrado de los cuatro.
- El **overfitting negativo de UMDA4GI a bajos presupuestos** en MiniSAT-hack es un resultado científicamente interesante: la distribución probabilística podría estar capturando propiedades estructurales del código que generalizan más allá del conjunto de entrenamiento.
- El **outlier de GP a budget=500 en MiniSAT-hack** (183.9%) debe mencionarse explícitamente en el paper. Es consistente con el comportamiento de GP de encontrar patches muy específicos del train.
- La coincidencia entre mejor mejora y menor overfitting para UMDA4GI refuerza la hipótesis de que el modelado probabilístico del espacio de búsqueda es beneficioso tanto para la exploración como para la generalización.

### Limitaciones

- `validate_point_or_zero`: celdas con folds incompletos devuelven overfit=0, subestimando potencialmente el overfitting real en esos casos (dos celdas afectadas: pngOptim/GP/1000 y pngOptim/UMDA4GI/750).
- El overfitting en Sat4J/Weka/OptimPNG es tan pequeño que el ranking allí tiene poca relevancia práctica.
